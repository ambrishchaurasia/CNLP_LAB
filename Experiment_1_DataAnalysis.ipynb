```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("/content/drive/MyDrive/CNLP-LAB/Experiment-1/employee_information_100.csv")
```

```python
avg_salary=df.groupby('Department')['Salary'].mean()
print(avg_salary)
```

```python
avg_salary.plot(kind='bar',color='green')
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.title('Average Salary by Department')
plt.show
```

#Experiment 1.2: Display the number of employees working in each department.


```python
dept_count=df['Department'].value_counts()
print(dept_count)
```

```python
dept_count.plot(kind='bar',color='teal')
plt.title("Number of employees in each department")
plt.xlabel("Department")
plt.ylabel("Number of employees")
plt.show()
```

#Experiment 1.3:Display the percentage of male and female employees using a pie chart


```python
gender_percentage = df["Gender"].value_counts(normalize=True) * 100

plt.pie(
    gender_percentage,
    labels=gender_percentage.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Percentage of Male and Female Employees")
plt.show()
```

Experiment 1.4: Draw a histogram showing salary distribution.

```python
plt.hist(df["Salary"], bins=8, edgecolor="black")

plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.title("Salary Distribution")
plt.show()
```

Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.

```python
plt.scatter(df["Experience_Years"], df["Salary"])

plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.title("Relationship Between Experience and Salary")
plt.show()
```

Experiment 1.6: Display the top 10 employees with the highest salary.

```python
top_10 = df.nlargest(10, "Salary")

print(top_10[["Employee_ID", "Name", "Department", "Salary"]])
```

Experiment 1.7: Find the highest salary in every department.

```python
highest_salary = df.groupby("Department")["Salary"].max()

print(highest_salary)
```


Experiment 1.8: Display employees whose salary is greater than the overall average salary.


```python
average_salary = df["Salary"].mean()

print("Overall Average Salary:", average_salary)

above_average = df[df["Salary"] > average_salary]

print(above_average[["Employee_ID", "Name", "Salary"]])
```

Experiment 1.9: Calculate the average years of experience in each department.

```python
average_experience = df.groupby("Department")["Experience_Years"].mean()

print(average_experience)
```

Experiment 1.10: Display the distribution of employee ages using a histogram.

```python
plt.hist(df["Age"], bins=8, edgecolor="black")

plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.title("Distribution of Employee Ages")
plt.show()
```
