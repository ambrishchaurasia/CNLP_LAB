

---



```python
upperc=0
for ch in text:
  if ch.isupper():
    upperc+=1

print("1. Uppercase characters:", upperc)

text=text.lower()
```

Remove punctuation.        (Also display how many different types of punctuations?)


```python
import string
pun=set()

for ch in text:
  if ch in string.punctuation:
    pun.add(ch)

print(len(pun))
```

```python
print(pun)
```

```python
for ch in pun:
  text=text.replace(ch,"")
```

```python
print(text)
```

```python
num=list()
for ch in text:
  if ch.isdigit():
    num.append(ch)
print(len(num))
print(num)
```

Remove extra whitespace. (How much extra whitespace is present?)


```python
for ch in num:
  text=text.replace(ch,"")
print(text)
```

```python
count=0

for i in range (0,len(text)-1):
  if(text[i]==" " and text[i+1]==" "):
    count+=1
print(count)
```

```python
text = " ".join(text.split())
```

Display the cleaned text.     


```python
print(text)
```

Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords



tokens = word_tokenize(text)


stop_words = set(stopwords.words("english"))


found_stop_words = []


filtered_tokens = []

for word in tokens:

    if word.lower() in stop_words:
        found_stop_words.append(word)

    else:
        filtered_tokens.append(word)


print("Original Tokens:")
print(tokens)

print("\nStop Words:")
print(found_stop_words)

print("\nFiltered Tokens:")
print(filtered_tokens)
```
